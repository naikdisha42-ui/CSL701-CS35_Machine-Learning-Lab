### Handling Missing Values in 'Age' Column

Let's fill the missing values in the 'Age' column with the mean age. This is a common strategy for numerical features, especially when the missingness is considered random.

```python
# Calculate the mean of the 'Age' column
mean_age = df_train['Age'].mean()
print(f"Mean Age: {mean_age:.2f}")

# Fill missing 'Age' values with the calculated mean
df_train['Age'].fillna(mean_age, inplace=True)

# Verify that there are no more missing values in the 'Age' column
print('\nMissing values in Age column after imputation:')
display(df_train['Age'].isnull().sum())
```

### Calculate Survival Rate by Sex

Let's calculate the survival rate for each gender to see if there's a significant difference in survival based on sex.

```python
# Calculate survival rate grouped by Sex
survival_rate_by_sex = df_train.groupby('Sex')['Survived'].mean().reset_index()

print('Survival Rate by Sex:')
display(survival_rate_by_sex)
```

### Handling Missing Values in 'Age' Column (with Median)

As previously noted, the missing values in the 'Age' column were already filled with the mean. This operation will calculate the median and attempt to fill any remaining missing values, though it won't change the DataFrame's 'Age' column since there are no `NaN` values left.

```python
# Calculate the median of the 'Age' column
median_age = df_train['Age'].median()
print(f"Median Age: {median_age:.2f}")

# Fill missing 'Age' values with the calculated median
df_train['Age'].fillna(median_age, inplace=True)

# Verify that there are no more missing values in the 'Age' column
print('\nMissing values in Age column after imputation with median:')
display(df_train['Age'].isnull().sum())
```

Task 13: Plot histogram of Age.

```python
# Plot a histogram of the 'Age' column
plt.figure(figsize=(10, 6))
sns.histplot(df_train['Age'], bins=20, kde=True)
plt.title('Distribution of Age on the Titanic')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

Task 14: Plot bar chart of Survival Count.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a countplot of 'Survived' column
plt.figure(figsize=(8, 6))
sns.countplot(x='Survived', data=df_train, palette='viridis', hue='Survived', legend=False)
plt.title('Survival Count (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.xticks(ticks=[0, 1], labels=['Not Survived', 'Survived'])
plt.grid(axis='y', alpha=0.75)
plt.show()
```

Task 15: Plot scatter graph of Age vs Fare.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=df_train, alpha=0.6)
plt.title('Scatter Plot of Age vs. Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
